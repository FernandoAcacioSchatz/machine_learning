1. O que significa GIGO? 
GIGO significa Garbage In, Garbage Out ("Lixo entra, lixo sai"). Isso quer dizer que, se os dados de entrada forem incorretos ou .de baixa qualidade, os resultados obtidos também serão ruins.

2. O que é um dataset? 
Um dataset é um conjunto de dados organizado. Ele pode estar em formatos como CSV, Excel, banco de dados ou JSON e é utilizado para análise, estatística e treinamento de modelos de Machine Learning.

3. O que é um DataFrame? 
Um DataFrame é uma estrutura de dados bidimensional da biblioteca Pandas. Ele se parece com uma tabela, contendo linhas e colunas, onde cada coluna pode armazenar um tipo de dado diferente.

4. Para que serve o Pandas? 
O Pandas é uma biblioteca do Python utilizada para manipulação e análise de dados

8. O que é uma biblioteca em Python?
Uma biblioteca em Python é um conjunto de códigos prontos para resolver tarefas específicas.

9. Qual é a função do Pandas?
Manipulação de dados


10. O que significa a palavra `import`?
importar a lib para o projeto.

11. Por que utilizamos `as pd`?
Alias ou apelido para faciliar o chamado da lib.

12. O que aconteceria se tentássemos usar `pd.DataFrame()` antes de importar o Pandas?
Erro. Python não vai entender que pd.DataFrame() é uma chamada para lib pandas.
